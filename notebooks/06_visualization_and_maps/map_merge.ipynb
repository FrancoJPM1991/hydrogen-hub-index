```python
import cv2
from pyprojroot import here
```

```python
RESULTS_DATA_DIR = here() / "results"
```

```python
LISA_relm = cv2.imread(RESULTS_DATA_DIR/'maps/LISA_RELM_map.png')
LISA_relms = cv2.imread(RESULTS_DATA_DIR/'maps/LISA_RELMS_map.png')

RELM = cv2.imread(RESULTS_DATA_DIR/'maps/RELM_heatmap1.png')
RELMS = cv2.imread(RESULTS_DATA_DIR/'maps/RELMS_heatmap1.png')
```

```python
LISA_RELMvsRELMS = cv2.hconcat([LISA_relm, LISA_relms])
RELMvsRELMS = cv2.hconcat([RELM, RELMS])
```

```python
cv2.imwrite(RESULTS_DATA_DIR/'maps/LISA_RELMvsRELMS.png', LISA_RELMvsRELMS)
cv2.imwrite(RESULTS_DATA_DIR/'maps/RELMvsRELMS.png', RELMvsRELMS)
```
