```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np
import pandas as pd
from pyprojroot import here
```

```python
INTERIM_DATA_DIR = here() / "data" / "interim"
```

```python
merge = pd.read_csv(INTERIM_DATA_DIR/'master_merged.csv')
summary = pd.read_csv(INTERIM_DATA_DIR/'master_summary.csv')
```

```python
# We transform all variables with a skewness greater than 1
log1p_vars = summary.loc[summary['skewness'] > 1, 'Unnamed: 0'].tolist()
binary_vars = ['highways, rails']
```

```python
df_norm = merge.copy()
```

```python
for col in log1p_vars:
    df_norm[col] = np.log1p(df_norm[col])

scaler = MinMaxScaler()
all_continuous = [col for col in summary['Unnamed: 0'].unique() if col not in binary_vars]
df_norm[all_continuous] = scaler.fit_transform(df_norm[all_continuous])
```

```python
invert_vars = [
    'distance_ciit',
    'dist_to_border_km',
    'dist_min_ports',
    'dist_min_km',
    'gen_min_km',
    'crime'
]
```

```python
result = df_norm.copy()
```

```python
for col in invert_vars:
    if col in invert_vars:
        result[col] = 1 - df_norm[col]
    else:
        result[col] = df_norm[col]
```

```python
result.to_csv(INTERIM_DATA_DIR/"master_norm.csv", encoding="utf-8")

print(f"Done. {len(df_norm)} municipalities written.")
print(result.head(10))
```
