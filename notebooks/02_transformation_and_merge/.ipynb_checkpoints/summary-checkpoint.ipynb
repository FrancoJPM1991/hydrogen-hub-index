```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
df = pd.read_csv('master_merged.csv')
```

```python
num_df = df.select_dtypes(include='number')

summary = pd.DataFrame({
    'min':      num_df.min(),
    'max':      num_df.max(),
    'mean':     num_df.mean(),
    'median':   num_df.median(),
    'std':      num_df.std(),
    'skewness': num_df.skew(),
    'kurtosis': num_df.kurt(),
    'missing':  num_df.isna().sum(),
})
```

```python
print(summary.to_string())
summary.to_csv("master_summary.csv", encoding="utf-8")
```
