```python
import pandas as pd
from pyprojroot import here
```

```python
INTERIM_DATA_DIR = here() / "data" / "interim"
PROCESSED_DATA_DIR = here() / "data" / "processed"
RESULTS_DATA_DIR = here() / "results" / "esda"
```

```python
df = pd.read_csv(RESULTS_DATA_DIR/"LISA_RELMS.csv", dtype={'CVEGEO': str})
rank = pd.read_csv(PROCESSED_DATA_DIR/"RELMS_index.csv", dtype={'CVEGEO': str})
helax_mun = "20014" # Ciudad Ixtepec, Oaxaca
```

```python
helax_rank = rank[rank['CVEGEO'] == helax_mun]
helax_class = df[df['CVEGEO'] == helax_mun].copy()
```

```python
helax_class.columns
```

```python
print("=" * 60)
print("HELAX green hydrogen plant municipality RELMS ranking and cluster")
print("=" * 60)
print(f" The helax plant municipality number is {helax_mun} and rank in the place number {helax_rank['rank']}.")
print(f" Likewise, such municipality was classified by the local Moran's I as follows:")
print(f" Quadrant: {helax_class['quadrant']}")
print(f" Cluster: {helax_class['cluster']}")
```
