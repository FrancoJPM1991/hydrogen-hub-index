```python
import geopandas as gpd
import pandas as pd
from pyprojroot import here
```

```python
RAW_DATA_DIR = here() / "data" / "raw"
INTERIM_DATA_DIR = here() / "data" / "interim"
```

```python
mun = gpd.read_file(RAW_DATA_DIR/"geography/municipalities/00mun.shp").to_crs(epsg=4326)
```

```python
# The crs warning is resolved by transforming it into centroids_wgs84.
mun["centroid"] = mun.geometry.centroid
centroids = gpd.GeoDataFrame(mun[["CVEGEO"]], geometry=mun["centroid"], crs=mun.crs)
centroids_wgs84 = centroids.to_crs(epsg=4326)
centroids_wgs84["long"] = centroids_wgs84.geometry.x
centroids_wgs84["lat"]  = centroids_wgs84.geometry.y
```

```python
out = centroids_wgs84[["CVEGEO", "long", "lat"]]
out.to_csv(INTERIM_DATA_DIR/"municipality_centroids.csv", index=False)
print(f"Done — {len(out)} municipalities exported to municipality_centroids.csv")
```
