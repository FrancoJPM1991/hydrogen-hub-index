```python
import pandas as pd
from pyprojroot import here
```

```python
RAW_DATA_DIR = here() / "data" / "raw"
INTERIM_DATA_DIR = here() / "data" / "interim"
```

```python
sec = pd.read_csv(RAW_DATA_DIR/"security/crime_rates_csv/Municipal-Delitos-2015-2025_abr2026.csv", encoding='latin-1')
```

We considered the following crimes as relevant:
- Robo a transportista: truck hijacking/cargo theft.
- Robo de vehículo automotor: car theft/carjacking.
- Robo de maquinaria: theft of machinery or equipment.
- Daño a la propiedad: property damage.
- Otros robos: other thefts

```python
crime_list = ['Robo a transportista', 'Robo de vehículo automotor', 'Robo de maquinaria', 'Daño a la propiedad', 'Otros robos']
crime = sec[sec['Subtipo de delito'].isin(crime_list)].copy()
```

```python
crime['total'] = crime.iloc[:, 9:21].sum(axis=1)
crime.columns.tolist()
```

```python
crime.drop(crime.iloc[:, 4:21], axis=1, inplace=True)
crime.drop(columns=['Entidad', 'Clave_Ent'], inplace=True)
```

```python
crime.rename(columns={'Cve. Municipio': 'CVEGEO', 'Año': 'year', 'total': 'crime'}, inplace=True)
crime['CVEGEO'] = crime['CVEGEO'].astype(str).str.zfill(5)
crime = crime[crime['year'] == 2025]
result = crime.groupby('CVEGEO')[['crime']].sum().reset_index()
```

```python
result.to_csv(INTERIM_DATA_DIR/"crime_by_municipality.csv")

print(f"Done. {len(result)} municipalities written.")
print(result.sample(10))
```
