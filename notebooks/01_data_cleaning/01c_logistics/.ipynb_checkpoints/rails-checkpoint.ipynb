```python
import pandas as pd
import geopandas as gpd
from pyprojroot import here
```

```python
RAW_DATA_DIR = here() / "data" / "raw"
INTERIM_DATA_DIR = here() / "data" / "interim"
```

```python
mun = gpd.read_file(RAW_DATA_DIR/"geography/municipalities/00mun.shp")
rails = gpd.read_file(RAW_DATA_DIR/"logistics/railways_shp/via_ferrea.shp")
```

```python
rails = rails.to_crs(mun.crs)
```

```python
join = gpd.sjoin(mun[["CVEGEO", "geometry"]], rails[["geometry"]], how="left", predicate="intersects")
```

```python
mun["rails"] = mun.index.map(
    join.groupby(join.index)["index_right"].apply(lambda x: int(x.notna().any()))
)
```

```python
result = mun[["CVEGEO", "rails"]]
result.to_csv(INTERIM_DATA_DIR/"rails_by_municipality.csv", index=False)

print(f"Done. {len(result)} municipalities written.")
print(result.sample(10))
```
