```python
import pandas as pd
from pyprojroot import here
```

```python
RAW_DATA_DIR = here() / "data" / "raw"
INTERIM_DATA_DIR = here() / "data" / "interim"
```

```python
eco_output_raw = pd.read_csv(RAW_DATA_DIR/"market/economic_output_xlsx/SAIC_Exporta_2026617_233048143.csv", 
                         encoding="utf-8",
                        dtype={0: str}
                        )
```

```python
eco_output_raw['ENT'] = eco_output_raw['Entidad'].str.extract(r'(\d+)').astype(str)
eco_output_raw['MUN'] = eco_output_raw['Municipio'].str.extract(r'(\d+)').astype(str)
eco_output_raw['CVEGEO'] = eco_output_raw['ENT'] + eco_output_raw['MUN']
```

```python
eco_output_raw = eco_output_raw[eco_output_raw["Año Censal"] == '2023'].copy()
eco_output_raw = eco_output_raw[eco_output_raw["Actividad económica"].str.contains('31-33')].copy()
```

```python
# A111A... contains the brute production output in millions of Mexican pesos.
eco_output = eco_output_raw.groupby('CVEGEO')[['A111A Producción bruta total (millones de pesos)']].sum()
eco_output.rename(columns={'A111A Producción bruta total (millones de pesos)': 'economic_output'}, inplace=True)
```

```python
eco_output.to_csv(INTERIM_DATA_DIR/"economic_output.csv", encoding="utf-8")

print(f"Done. {len(eco_output)} municipalities written.")
print(eco_output.sample(10))
```
