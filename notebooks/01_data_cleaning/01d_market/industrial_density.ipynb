```python
import pandas as pd
import geopandas as gpd
from pyprojroot import here
```

```python
RAW_DATA_DIR = here() / "data" / "raw"
INTERIM_DATA_DIR = here() / "data" / "interim"
```

```python
industry_hubs = gpd.read_file(RAW_DATA_DIR/"market/economic_units_locations_shp/denue_inegi_31-33_.shp")
mun = gpd.read_file(RAW_DATA_DIR/"geography/municipalities/00mun.shp")
```

```python
industry_hubs['CVEGEO'] = industry_hubs['cve_ent'].astype(str).str.zfill(2) + industry_hubs['cve_mun'].astype(str).str.zfill(3)
```

```python
industry_count = (industry_hubs.groupby('CVEGEO')
                .size()
                .reset_index(name='n_industy'))
```

```python
industrial_density = mun.merge(industry_count, on='CVEGEO', how='left')       
```

```python
industrial_density['n_industy'] = industrial_density['n_industy'].fillna(0).astype(int)
result = industrial_density.drop(columns=['geometry', 'NOMGEO', 'CVE_ENT', 'CVE_MUN'])
```

```python
result.to_csv(INTERIM_DATA_DIR/"industrial_density.csv", index=False)

print(f"Done. {len(result)} municipalities written.")
print(result.sample(10))
```
