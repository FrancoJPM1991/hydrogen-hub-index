```python
import geopandas as gpd
import pandas as pd
from pyprojroot import here
```

```python
RAW_DATA_DIR = here() / "data" / "raw"
INTERIM_DATA_DIR = here() / "data" / "interim"
```

```python
mun = gpd.read_file(RAW_DATA_DIR/"geography/municipalities/00mun.shp")
water = gpd.read_file(RAW_DATA_DIR/"resources/water_reservoirs_shp/disponibilidad_de_agua_subterranea_09-11-2023.shp")
```

```python
# We verify that both gpd share the same CRS
if mun.crs != water.crs:
    water = water.to_crs(mun.crs)
```

```python
# DMA_POSITI is the total available subterranean water in cubic million meters.
joined = gpd.sjoin(mun, water[["DMA_POSITI", "geometry"]], how="left", predicate="intersects")
```

```python
result = joined[["CVEGEO", "DMA_POSITI"]].drop_duplicates(subset="CVEGEO")
```

```python
result.rename(columns={'DMA_POSITI': 'water_availability'}, inplace=True)
```

```python
result.to_csv(INTERIM_DATA_DIR/"water_availability_by_municipality.csv", index=False)

print(f"Done. {len(result)} municipalities written.")
print(result.sample(10))
```
